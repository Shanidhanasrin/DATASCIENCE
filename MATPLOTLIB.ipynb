```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[2,4,6,8,10]

plt.plot(x,y)   #line chart
plt.show()    #display print
```

```python
month=['jan','feb','mar','apr','may']
sales=[20,30,25,40,50]

plt.plot(month,sales)
plt.show()
```

```python
month=['jan','feb','mar','apr','may']
sales=[20,30,25,40,50]
profit=[5,10,15,20,25]

plt.plot(month,sales,label="sales")  #label means name the label
plt.plot(month,profit,label="profit")
plt.legend()  #display the names
plt.xlabel("month")
plt.ylabel("sales")
plt.title("montly sales")
plt.grid(True) #to grid
plt.plot(
    month,
    sales,
    marker="*", #show individual data points  (o=circle,s=square)
    color="green",  #to change the color
    linestyle="--"
)
plt.show()
```

```python
#bar chart
month=['jan','feb','mar','apr','may']
sales=[20,30,25,40,50]
profit=[5,10,15,20,25]

plt.xlabel("month")
plt.ylabel("sales")
plt.title("montly sales")
plt.grid(True) #to grid
plt.bar(
    month,
    sales,
    color="green"
)
plt.show()
```

```python
#scatter
study_hours=[1,2,3,4,5,6]
marks=[40,45,55,65,75,85]
plt.scatter(study_hours,marks)
plt.title("study hours vs marks")
plt.xlabel("study hours")
plt.ylabel("marks")
plt.show()
```

```python
#histogram
marks=[30,22,15,28,45,65,78,89,43,32,12,76,65]
plt.hist(marks)
plt.title("stusy hour vs marks")
plt.xlabel("marks")
plt.ylabel("frequency")
plt.show()
```

```python
#pie chart
categories=["food","travel","shopping","others"]
amount=[2000,4000,3000,1000]
plt.pie(amount,labels=categories)
plt.title(" montly expenses")
plt.show()
```

```python
#2graph in 1
fig,ax=plt.subplots(2,1) #fig=whole graph ,ax=individual graph(2,1):row&column
ax[0].plot([1,2,3],[10,20,30])
ax[1].bar(["a","b","c"],[20,15,30])
plt.show()
```
