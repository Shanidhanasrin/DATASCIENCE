```python
import pandas as pd
data=[10,30,50,70]
s=pd.Series(data)      #convert into series,capital S
print(s)
```

```python
data=[10,30,50,70]
s=pd.Series(data,index=['a','b','c','d'])  #index=for changing index value
print(s)
```

```python
data={'name':'shan','age':18,'place':'valayamkulam'}
s=pd.Series(data)
print(s)
```

loc and iloc=accessing and slicing

iloc=default index based


loc=manual index based

```python
data=[20,30,40,50]
s=pd.Series(data,index=['a','b','c','d'])
print(s.loc['a'])
print(s.iloc[1])
```
