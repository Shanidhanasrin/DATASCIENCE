```python
import pandas as pd
data=[10,30,50,70]
s=pd.Series(data)      #convert into series,capital S
print(s)
```

```python
data=[10,30,50,70]
s=pd.Series(data,index=['a','b','c','d'])  #index=for changing index value
print(s)
```

```python
data={'name':'shan','age':18,'place':'valayamkulam'}
s=pd.Series(data)
print(s)
```

loc and iloc=accessing and slicing

iloc=default index based


loc=manual index based

```python
data=[20,30,40,50]
s=pd.Series(data,index=['a','b','c','d'])
print(s.loc['a'])  #LABEL BASED
print(s.iloc[1])    #INTIGER BASED
```

```python
#slicing
data=[30,20,40,50]
s=pd.Series(data,index=['a','b','c','d'])
print(s.loc['a':'b'])
print(s.iloc[:2])
```

```python
print(s)
print(s[s<30])
print(s[s>30])
print(s[s<=30])
print(s[s>=30])
print(s[s!=30])
```

```python
import pandas as pd
sales=pd.Series([1200,850,1500,950,2000],
                index=["monday","tuesday","wednesday","thursday","friday"])
print(sales)
print(sales.max())
print(sales.sum())
print(sales.min())
print(sales[sales<1000])
print(sales[sales>1000])
```

```python
import pandas as pd
data=[20,20,30,30,50,60,60,80]
ar=pd.Series(data)
print(ar)
print(ar.unique())  #unique for removing
print(ar.nunique()) # count
print(ar.value_counts()) #repeated value remove
```

```python
data=[10,20,None,40,50,50,None,60]
a=pd.Series(data)
print(a)
print(a.isnull())
print(a.notnull())
print(a.dropna())  #remove null value
```
