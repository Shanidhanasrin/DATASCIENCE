```python
#total
import numpy as np
marks=np.array([78,92,65,88,45,90,72,55])
print(np.sum(marks))
```

```python
#average
import numpy as np
marks=np.array([78,92,65,88,45,90,72,55])
print(np.average(marks))
```

```python
#highest mark
import numpy as np
marks=np.array([78,92,65,88,45,90,72,55])
print(np.max(marks))
```

```python
#lowest
import numpy as np
marks=np.array([78,92,65,88,45,90,72,55])
print(np.min(marks))
```

```python
#find student who score above 75
marks=np.array([78,92,65,88,45,90,72,55])
print(marks[marks>75])
```

```python
#below 50 and =50
marks=np.array([78,92,65,88,45,90,72,55])
print(marks[marks<=50])
```

```python
#sorting (ascenting order)
marks=np.array([78,92,65,88,45,90,72,55])
print(np.sort(marks))
```

```python
#find mean temperature
temperature=np.array([32,35,40,28,42,33])
print(np.mean(temperature))

```

```python
#find maximum and minimum
print(np.min(temperature))
print(np.max(temperature))
```

```python
#replace temperature above 38 with 38
temperature[temperature>38]=38                    #
print(temperature)
```

```python
#reshape the array into 2x3 matrix
print(temperature.reshape(2,3))
```

```python
#display the first row and second column values
temp=temperature.reshape(2,3)
print(temp)
print(temp[0])
print(temp[:,2])
```

```python
#slicing
c=[10,20,30,40,50,60,70,80]
print(c[0::7])
#print index 2 to 6
print(c[1:5])
#alternative element
print(c[::2])
```

```python
#even index
a=np.array([1,2,3,4,5,6,7,8])
print(a)
print(np.where(a%2==0))
#odd index
print(np.where(a%2!=0))
#normal
print(a[a%2==0])
print(a[a%2!=0])
```
