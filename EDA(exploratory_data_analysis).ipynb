```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv('/titanic.csv')
df


```

```python
print(df.shape) #to show no of columns and rows
```

```python
df.info() #complete details like column name,data types
```

```python
df.describe() #descriptive statstics
```

```python
df.isnull().sum() #tofind null values. if null value true then if no null value false
                  #sum= return true as & 1 false as 0
```

```python
df["Embarked"]=df["Embarked"].fillna(    #replace missing value
    df["Embarked"].mode()[0]    #find the most frequent value
)
print(df.isnull().sum())
print("before:",df.shape[0])   #0=ROW 1=COLUMNS

df=df.drop_duplicates()
print("after:",df.shape[0])
```

```python
df=df.rename(columns={
    "Passengerid":"PassengerID",
    "2urvived":"survived"
})
df
```

```python
df["Sex"]=df["Sex"].map({
    0:"male",
    1:"female"
})
df
```

```python
df.head()
sns.histplot(data=df,x="Age",bins=20)
```

```python
plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("number of passengers")
plt.show()
sns.countplot(data=df,x="Sex",hue="Survived")

plt.title("Titanic survival by sex")
plt.xlabel("Sex")
plt.ylabel("number of passengers")
plt.show()
sns.countplot(data=df,x="Pclass",hue="Survived")

plt.title("Titanic survival by  passenger class")
plt.xlabel("Passenger class")
plt.ylabel("number of passengers")
plt.show()
```
